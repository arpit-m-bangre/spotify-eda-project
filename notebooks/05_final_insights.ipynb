# 📌 Executive Summary

The Spotify EDA project explored relationships between track popularity, audio characteristics, emotional features, and genre-based trends.

The analysis revealed that track popularity is influenced by multiple interacting musical attributes rather than a single dominant feature.

Features such as danceability, energy, loudness, and genre-specific characteristics contribute differently to listener engagement patterns across Spotify’s music catalog.

# 📊 Key Findings

- Most Spotify tracks exhibit moderate-to-high danceability and energy levels.
- Popularity demonstrates weak direct correlation with individual audio features.
- High-popularity tracks tend to show slightly higher danceability and energy characteristics.
- Explicit tracks exhibit marginally higher median popularity levels compared to non-explicit tracks.
- Genres such as K-pop, pop-film, and chill achieved strong average popularity scores.
- Metal and electronic genres dominated energy rankings, while Latin and rhythm-focused genres demonstrated the highest danceability levels.
- Listener engagement appears influenced by combinations of rhythm, energy, emotional tone, and genre appeal.

# 💼 Business Insights

- Spotify listener engagement patterns favor energetic and rhythmically engaging tracks.
- Genre-specific audience behavior plays a significant role in track popularity outcomes.
- Emotionally positive tracks may provide slight engagement advantages, although emotional tone alone is not a dominant popularity factor.
- High diversity across genres and audio characteristics supports broad listener segmentation opportunities.
- Audio features such as danceability and energy can assist in playlist optimization and recommendation system improvements.

# 🚀 Recommendations

- Promote highly danceable and energetic tracks in mainstream playlist strategies.
- Utilize genre-specific recommendation systems to improve listener personalization.
- Combine multiple audio characteristics rather than relying on single-feature optimization when predicting track success.
- Explore mood-based playlist generation using valence and energy features.
- Further analysis using machine learning models could improve popularity prediction accuracy.
