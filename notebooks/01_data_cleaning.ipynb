```python
# =========================================
# IMPORTING REQUIRED LIBRARIES
# =========================================

# Data manipulation
import pandas as pd
import numpy as np

# Data visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Optional settings for better display
pd.set_option('display.max_columns', None)

print("Libraries Imported Successfully 🚀")
```

```python
# =========================================
# DOWNLOADING SPOTIFY DATASET FROM KAGGLE
# =========================================

import kagglehub

# Download dataset
path = kagglehub.dataset_download(
    "maharshipandya/-spotify-tracks-dataset"
)

print("Dataset downloaded successfully ✅")
print("Dataset path:", path)
```

```python
# =========================================
# CHECK FILES INSIDE DOWNLOADED DATASET
# =========================================

import os

files = os.listdir(path)

print(files)
```

```python
# =========================================
# COPY DATASET INTO PROJECT RAW DATA FOLDER
# =========================================

import shutil
import os

# Source file path
source_file = os.path.join(path, files[0])

# Destination path
destination_file = "../data/raw/spotify.csv"

# Copy file
shutil.copy(source_file, destination_file)

print("Dataset copied successfully 🚀")
```

```python
# =========================================
# LOAD DATASET
# =========================================

df = pd.read_csv("../data/raw/spotify.csv")

print("Dataset Loaded Successfully ✅")
```

# 🧹 Removing Unnecessary Columns

The column `Unnamed: 0` appears to be an automatically generated index column and does not contribute to analysis.

It will be removed to keep the dataset clean and efficient.

```python
# =========================================
# REMOVE UNNECESSARY COLUMN
# =========================================

df.drop(columns=['Unnamed: 0'], inplace=True)

print("Column Removed Successfully ✅")
```

```python
df.columns
```

# 🏷️ Feature Categorization

Separating numerical and categorical columns for easier exploratory data analysis.# 🏷️ Feature Categorization

Separating numerical and categorical columns for easier exploratory data analysis.

```python
# =========================================
# IDENTIFY NUMERICAL COLUMNS
# =========================================

numerical_columns = df.select_dtypes(include=['int64', 'float64']).columns

print("Numerical Columns:\n")
print(numerical_columns)
```

```python
# =========================================
# IDENTIFY CATEGORICAL COLUMNS
# =========================================

categorical_columns = df.select_dtypes(
    include=['object', 'string', 'bool']
).columns

print("Categorical Columns:\n")
print(categorical_columns)
```

```python
# =========================================
# SAVE CLEANED DATASET
# =========================================

df.to_csv("../data/cleaned/spotify_cleaned.csv", index=False)

print("Cleaned dataset saved successfully ✅")
```
