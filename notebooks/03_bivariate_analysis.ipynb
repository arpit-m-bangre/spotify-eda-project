# 📈 Spotify EDA — Bivariate Analysis

This notebook explores relationships between Spotify audio features and track popularity using comparative visual analysis.

```python
# =========================================
# IMPORT REQUIRED LIBRARIES
# =========================================

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

# Display all columns
pd.set_option('display.max_columns', None)

# Professional plot styling
sns.set_style("whitegrid")

print("Libraries Imported Successfully ✅")
```

```python
# =========================================
# LOAD CLEANED DATASET
# =========================================

df = pd.read_csv("../data/cleaned/spotify_cleaned.csv")

print("Dataset Loaded Successfully ✅")
```

```python
# Quick dataset preview

df.head()
```

# 🎵 Popularity vs Danceability

```python
# =========================================
# POPULARITY VS DANCEABILITY
# =========================================

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='danceability',
    y='popularity',
    alpha=0.5
)

plt.title("Popularity vs Danceability", fontsize=16)
plt.xlabel("Danceability")
plt.ylabel("Popularity")
# POPULARITY VS DANCEABILITY
plt.savefig(
    "../visuals/charts/popularity_vs_danceability.png",
    bbox_inches='tight'
)

plt.show()
```

### Observations

- Tracks with moderate-to-high danceability are widely represented across different popularity levels.
- Songs with extremely low danceability appear less common within the dataset.
- No strong linear relationship is observed between danceability and popularity, suggesting that popularity is influenced by multiple factors beyond danceability alone.
- Highly popular tracks are distributed across a broad range of danceability values, although moderately danceable tracks appear more frequently.
- The dense concentration of points indicates substantial variability in popularity even among tracks with similar danceability scores.

# 🔞 Explicit Content vs Popularity

```python
# =========================================
# EXPLICIT CONTENT VS POPULARITY
# =========================================

plt.figure(figsize=(8, 6))

sns.boxplot(
    data=df,
    x='explicit',
    y='popularity'
)

plt.title("Explicit Content vs Popularity", fontsize=16)
plt.xlabel("Explicit Content")
plt.ylabel("Popularity")
# EXPLICIT CONTENT VS POPULARITY
plt.savefig(
    "../visuals/charts/explicit_vs_popularity.png",
    bbox_inches='tight'
)
plt.show()
```

### Observations

- Explicit tracks appear to have a slightly higher median popularity compared to non-explicit tracks.
- Both explicit and non-explicit songs exhibit a broad popularity distribution across the dataset.
- The popularity spread for explicit tracks appears marginally wider, indicating greater variability in listener engagement.
- Non-explicit tracks contain a few high-popularity outliers approaching maximum popularity levels.
- Overall, the difference between explicit and non-explicit track popularity is noticeable but not extremely large.

# 🎼 Genre vs Popularity

```python
# =========================================
# TOP 10 GENRES BY AVERAGE POPULARITY
# =========================================

# Calculate average popularity by genre
genre_popularity = (
    df.groupby('track_genre')['popularity']
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

# Create figure
plt.figure(figsize=(12, 6))

# Create barplot
sns.barplot(
    x=genre_popularity.values,
    y=genre_popularity.index
)

plt.title("Top 10 Genres by Average Popularity", fontsize=16)
plt.xlabel("Average Popularity")
plt.ylabel("Genre")
# TOP 10 GENRES BY AVERAGE POPULARITY
plt.savefig(
    "../visuals/charts/top_genres_by_popularity.png",
    bbox_inches='tight'
)

plt.show()
```

### Observations

- Pop-film and K-pop genres exhibit the highest average popularity levels within the dataset.
- Several niche and mood-oriented genres such as chill, sad, and emo also demonstrate strong popularity performance.
- The differences in average popularity across the top genres are noticeable but not extremely large.
- Mainstream genres such as pop continue to maintain high popularity, although some specialized genres outperform them in this dataset.
- The results suggest that Spotify listener engagement is distributed across both mainstream and niche musical categories.

# 🔥 Correlation Heatmap

```python
# =========================================
# CORRELATION HEATMAP
# =========================================

# Select numerical columns
numerical_df = df.select_dtypes(include=['int64', 'float64'])

# Compute correlation matrix
corr_matrix = numerical_df.corr()

# Create figure
plt.figure(figsize=(14, 10))

# Generate heatmap
sns.heatmap(
    corr_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Heatmap of Spotify Features", fontsize=16)
# CORRELATION HEATMAP
plt.savefig(
    "../visuals/charts/correlation_heatmap.png",
    bbox_inches='tight'
)

plt.show()
```

### Observations

- Energy and loudness exhibit a strong positive correlation, indicating that louder tracks are generally more energetic.
- Energy and acousticness show a strong negative correlation, suggesting that highly acoustic tracks tend to have lower energy levels.
- Danceability demonstrates a moderate positive relationship with valence, indicating that more danceable songs often exhibit more positive emotional characteristics.
- Popularity shows relatively weak correlations with most individual audio features, implying that track success is influenced by multiple combined factors rather than a single characteristic.
- Instrumentalness displays negative correlations with both popularity and valence, suggesting that instrumental tracks may receive comparatively lower listener engagement and emotional positivity scores.

# 📌 Bivariate Analysis Summary

### Key Findings

- Danceability and energy show moderate positive relationships, indicating that energetic tracks are often rhythmically engaging.
- Popularity does not exhibit a strong linear relationship with any single audio feature.
- Explicit tracks demonstrate slightly higher median popularity levels compared to non-explicit tracks.
- Several niche genres such as K-pop, pop-film, and chill achieve strong average popularity performance.
- Correlation analysis reveals strong relationships between energy, loudness, and acousticness.

### Conclusion

The bivariate analysis suggests that Spotify track popularity is influenced by a combination of musical characteristics rather than a single dominant feature.

Audio features such as energy, danceability, loudness, and emotional tone contribute differently across genres and listener preferences, highlighting the complexity of music engagement patterns on streaming platforms.
