# 📊 Popularity Distribution Analysis

Analyzing the distribution of Spotify track popularity scores.

This helps identify:
- common popularity ranges,
- skewness,
- spread,
- and concentration of popular tracks.

```python
# =========================================
# POPULARITY DISTRIBUTION
# =========================================

plt.figure(figsize=(12, 6))

sns.histplot(
    df['popularity'],
    bins=30,
    kde=True
)

plt.title("Distribution of Track Popularity", fontsize=16)
plt.xlabel("Popularity")
plt.ylabel("Frequency")

plt.savefig(
    "../visuals/charts/popularity_distribution.png",
    bbox_inches='tight'
)

plt.show()
```

## 📌 Observations

- The popularity distribution is positively skewed, with a long tail extending toward highly popular tracks.
- A significant number of songs have very low popularity scores, indicating the presence of niche or less-streamed tracks.
- Most tracks are concentrated within the medium popularity range (20–60).
- Extremely popular songs are relatively rare, suggesting that only a limited portion of tracks achieve widespread listener engagement.
- The distribution does not follow a perfectly normal pattern and shows noticeable variation across popularity levels.

# 🎶 Danceability Analysis

Analyzing the distribution of danceability scores across Spotify tracks.

Danceability describes how suitable a track is for dancing based on:
- tempo,
- rhythm stability,
- beat strength,
- and overall musical regularity.

```python
# =========================================
# DANCEABILITY DISTRIBUTION
# =========================================

plt.figure(figsize=(12, 6))

sns.histplot(
    df['danceability'],
    bins=30,
    kde=True
)

plt.title("Distribution of Danceability", fontsize=16)
plt.xlabel("Danceability")
plt.ylabel("Frequency")
# DANCEABILITY DISTRIBUTION
plt.savefig(
    "../visuals/charts/danceability_distribution.png",
    bbox_inches='tight'
)
plt.show()
```

```python
# =========================================
# DANCEABILITY BOXPLOT
# =========================================

plt.figure(figsize=(10, 2))

sns.boxplot(
    x=df['danceability']
)

plt.title("Boxplot of Danceability", fontsize=14)
# DANCEABILITY BOXPLOT
plt.savefig(
    "../visuals/charts/danceability_boxplot.png",
    bbox_inches='tight'
)

plt.show()
```

## 📌 Observations

- The danceability distribution appears relatively close to a normal distribution with a concentration around moderate-to-high values.
- Most Spotify tracks exhibit danceability scores between 0.5 and 0.8, suggesting a strong preference for rhythmically engaging music.
- Extremely low danceability tracks are comparatively rare within the dataset.
- The boxplot indicates the presence of a few lower-end outliers representing tracks with unusually low danceability characteristics.
- Overall, the dataset suggests that highly danceable tracks are common across Spotify’s music catalog.

# 🎼 Track Genre Analysis

Analyzing the distribution of music genres within the Spotify dataset.

This analysis helps identify:
- dominant genres,
- genre diversity,
- and representation patterns across the platform.

To improve readability, only the top 15 most common genres are visualized.

```python
# =========================================
# TOP 15 MOST COMMON TRACK GENRES
# =========================================

# Get top 15 genres
top_genres = df['track_genre'].value_counts().head(15)

# Create figure
plt.figure(figsize=(14, 7))

# Create barplot
sns.barplot(
    x=top_genres.values,
    y=top_genres.index
)

# Titles and labels
plt.title("Top 15 Most Common Track Genres", fontsize=16)
plt.xlabel("Number of Tracks")
plt.ylabel("Genre")
# TOP 15 TRACK GENRES
plt.savefig(
    "../visuals/charts/top_15_track_genres.png",
    bbox_inches='tight'
)
# Show plot
plt.show()
```

## 📌 Observations

- The top genres appear to have nearly identical track counts, with each genre containing approximately the same number of songs.
- This suggests that the dataset may have been intentionally balanced across genres during collection or preprocessing.
- The balanced distribution improves fairness for comparative genre analysis.
- The dataset includes a wide variety of genres, indicating strong genre diversity within the Spotify catalog.
- Since genre representation is relatively uniform, future analyses are less likely to be biased toward a single dominant genre.

# 🔞 Explicit Content Analysis

Analyzing the distribution of explicit and non-explicit tracks within the Spotify dataset.

This analysis helps understand:
- content labeling trends,
- prevalence of explicit music,
- and listener-oriented content patterns.

```python
# =========================================
# EXPLICIT CONTENT DISTRIBUTION
# =========================================

plt.figure(figsize=(8, 5))

sns.countplot(
    x='explicit',
    data=df
)

plt.title("Distribution of Explicit Content", fontsize=16)
plt.xlabel("Explicit Content")
plt.ylabel("Number of Tracks")
# EXPLICIT CONTENT DISTRIBUTION
plt.savefig(
    "../visuals/charts/explicit_content_distribution.png",
    bbox_inches='tight'
)

plt.show()
```

```python
# =========================================
# EXPLICIT CONTENT PERCENTAGE
# =========================================

explicit_percentage = (
    df['explicit']
    .value_counts(normalize=True) * 100
)

print(explicit_percentage)
```

## 📌 Observations

- Non-explicit tracks overwhelmingly dominate the dataset.
- Approximately 91% of tracks are labeled as non-explicit, while explicit songs represent a relatively small proportion of the catalog.
- This imbalance may reflect Spotify’s emphasis on broader audience accessibility and mainstream listener preferences.
- The explicit content variable is highly imbalanced, which may influence future comparative or predictive analyses.
- Overall, the dataset suggests that non-explicit music remains significantly more prevalent across Spotify tracks.
